```python
from fastai.vision.all import *
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from functools import partial
from numpy.random import normal,seed, uniform
```

```python
# Quadratic function 
def quadratic(x):
    return x**2
#plotting using matplotlib
x = np.linspace(-2, 2, 400)
y = quadratic(x)
plt.plot(x, y)
plt.xlabel('x')
plt.ylabel('quadratic(x)')
plt.title('Quadratic Function')
#point in the quadratic function
plt.scatter(-1.5, quadratic(-1.5), color='red')
plt.show()
#calculating the gradient  using tensor
def quadratic_for_tensor(x):
    x_tensor = tensor(x).requires_grad_()
    y_tensor = (x_tensor ** 2)
    y_tensor.backward()
    return x_tensor.grad
#plotting the gradient from -2 to 2 value of x
x_val = np.linspace(-2,2,400)
y_val = [quadratic_for_tensor(x) for x in x_val]
plt.plot(x_val, y_val)
plt.xlabel('x')
plt.ylabel('gradient of quadratic(x)')
plt.title('Gradient of Quadratic Function')
plt.show()
# adding sum to y_tensor 
def quadratic_sum_for_tensor(x):
    x_tensor = tensor(x).requires_grad_()
    y_tensor = (x_tensor ** 2).sum()
    y_tensor.backward()
    return x_tensor.grad
```

```python
# Measuring the speed of roler coster and then calculating the gradient using tensors
#Time using torch arrange from 0 to 20 seconds
time = torch.arange(0, 20).float()
#randomly assigning speeds as accurate speed measurement
speed = torch.randn(20)*3 + 0.75*(time-9.5)**2 + 1
plt.scatter(time,speed);
def f(t, params):
    a, b, c = params
    return a*t**2 + b*t + c
def mse(pred, target):
    return ((pred - target)**2).mean()
params = torch.randn(3).requires_grad_()
preds = f(time, params)
#graph to show the difference between predicted and actual value
def show_def(preds, ax=None):
    if ax is None:
        ax = plt.gca()
    ax.scatter(time, speed, label='Actual')
    ax.plot(time, preds.detach(), color='red', label='Predicted')
    ax.set_xlabel('Time')
    ax.set_ylabel('Speed')
    ax.set_title('Predicted vs Actual Speed')
    ax.legend()
    plt.show()

show_def(preds)
lr = 1e-5
# updating the parameters using gradient descent
def apply(params,prn=True): 
  preds = f(time, params)
  loss =mse(preds,speed)
  loss.backward()
  params.data-= lr * params.grad.data
  params.grad = None
  if(prn): print(loss.item())
  return preds
for i in range(2):
  apply(params)
 
```
